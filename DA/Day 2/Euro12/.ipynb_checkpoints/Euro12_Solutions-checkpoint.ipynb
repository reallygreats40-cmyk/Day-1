```python
#  a sample DataFrame
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Emma'],
    'Age': [25, 30, 35, 40, 45],
    'City': ['New York', 'Boston', 'Chicago', 'Denver', 'Seattle'],
    'Salary': [60000, 70000, 80000, 90000, 100000]
}

df = pd.DataFrame(data)

# Set a custom index
df.index = ['a', 'b', 'c', 'd', 'e']

print("Original DataFrame:")
df
```

**`loc` (Label-based indexing)**

- loc uses the labels of the index for selection. It's used for selecting data based on the label of the rows and columns.

```python
# Select a single row by label
print("\nSelecting row 'c' using loc:")
print(df.loc['c'])
```

```python
# Select multiple rows by label
print("\nSelecting rows 'a', 'c', and 'e' using loc:")
print(df.loc[['a', 'c', 'e']])
```

```python
# Select specific rows and columns
print("\nSelecting rows 'b' to 'd' and columns 'Name' and 'Salary' using loc:")
print(df.loc['b':'d', ['Name', 'Salary']])

```

```python
# Boolean indexing with loc
print("\nSelecting rows where Age > 30 using loc:")
print(df.loc[df['Age'] > 30])
```

**`iloc` (Integer-based indexing)**

- iloc uses integer positions for selection. It's used for selecting data based on the position of the rows and columns.

```python
df
```

```python
# Select a single row by position
print("\nSelecting the 3rd row (index 2) using iloc:")
print(df.iloc[2])
```

```python
# Select multiple rows by position
print("\nSelecting rows 0, 2, and 4 using iloc:")
print(df.iloc[[0, 2, 4]])

# Select specific rows and columns by position
print("\nSelecting rows 1 to 3 and columns 0 and 3 using iloc:")
print(df.iloc[1:4, [0, 3]])

# Slicing with iloc
print("\nSelecting all rows and columns 1 to 3 using iloc:")
print(df.iloc[:, 1:4])
```

## Key difference:

- `loc` uses labels (index values), while iloc uses integer positions
