## Key dataframe operations
### selecting, sorting, adding and aggregating data in dataframes

```python
import csv

def create_csv_files(data_list, filename):
    for i in range(len(data_list)):
        a_file = f"{filename}{i+1}.csv"
        with open(a_file, 'w') as file:
            writer = csv.writer(file)
            writer.writerow(data_list[i])

# Example usage with 10 records
students_data = [
    ['John', 25, 'Engineer'],
    ['Mary', 30, 'Doctor'],
    ['Peter', 28, 'Teacher'],
    ['Sarah', 35, 'Nurse'],
    ['Mike', 22, 'Student'],
    ['Lisa', 29, 'Designer'],
    ['Tom', 31, 'Manager'],
    ['Anna', 26, 'Writer'],
    ['David', 33, 'Analyst'],
    ['Emma', 27, 'Developer']
]

# for i in range(len(students_data)):
#     print(students_data[i])

# Create CSV files
create_csv_files(students_data, 'students')
```
