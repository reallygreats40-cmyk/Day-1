Consider the simple factorial function in Maths. Factorial of `5! = 5x4x3x2x1`; and the factorial of `7! = 7x6x5x4x3x2x1`. <br>

#### Task A: On lists
For the list `data = [23, 12, 9, 102, 89, 210]`:<br>
- write a python function called `calc_total()`, pass it a list as a parameter
- use either a while loop or a for loop to compute the total value of the list items

```python
# solve it here
data = [23, 12, 9, 102, 89, 210]

def calc_total(my_list):
    total = 0
    for i in my_list:
        total = total + i
    print(total)

calc_total(data)
    
```

```python
def calc_total2(my_list):
    total = 0
    s = 0
    while s < len(my_list):
        total = total + my_list[s]
        s = s + 1
    return total

calc_total2(data)
```

## Activity 5: Data Persistence


#### Task A: writing content to a text file
Consider the lists in Task C above: <br>  
account_vals = [123,20,402,300,65] <br>
account_names = ["Amy", "Mike", "Tina", "Abdullah", "Jane"] <br>

1. define a and open file called `people_data` and associate it to a disk file called `people.txt`. The file is to be written to.
2. Use a for loop to write contents of the lists to the text file. Use the following format - `name, account value`, line by line
3. Remember to close the file.
4. Check that the file `people.txt` has been created and has the contents expected

```python
# solve Task A here

account_vals = [123,20,402,300,65]
account_names = ["Amy", "Mike", "Tina", "Abdullah", "Jane"]

people_data = open("people.txt", 'w')  # open file
for i in range(len(account_names)): # iterate through the lists
    people_data.write(account_names[i]+","+str(account_vals[i])+"\n") # write elements separated by a comma
people_data.close() # close file
```
