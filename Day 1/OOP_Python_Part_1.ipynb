## 2. From functions and data to objects

Python already uses objects everywhere:
- `10` is an object (an `int`).
- `"hello"` is an object (a `str`).
- `{"UK": "London"}` is an object (a `dict`).

Each of these bundles **data** with **behaviour** (methods like `.upper()` or `.append()`). A class lets us define our *own* such bundles.

````python
# Python's built-in values are objects too.
city = {"UK": "London", "KE": "Nairobi"}

print(type(city))     # <class 'dict'> -- dict is a class; city is an object of that class
print(city.keys())    # .keys() is a method the dict class provides
````

### Why use a class?

A class lets us keep related **data + behaviour** together.

Instead of separate variables and standalone functions passed around, a class groups a `Person`'s `name` and `age` with the actions that make sense for a person, in one reusable definition.

## 3. Defining a class and creating objects

The `class` statement defines the blueprint. `__init__()` runs automatically whenever a new object is created, and sets up that object's initial state.

````python
class Person:
    """Represent a person with a name and age."""

    def __init__(self, name, age):
        """Initialise the person's name and age."""
        self.name = name
        self.age = age


person1 = Person("Amina", 30)
person2 = Person("David", 25)

print(person1.name, person1.age)
print(person2.name, person2.age)
````

### Class vs object

```text
        Person class
       /            \
      /              \
 person1            person2
 name = Amina        name = David
 age  = 30           age  = 25
```

One class, many independent objects. Each object has its **own** copy of the attributes.

````python
# Manipulating one object changes that object's state only.
person1.age = 31

print("person1:", person1.name, person1.age)
print("person2:", person2.name, person2.age)   # unaffected
````

## 4. Attributes, `self` and methods

### Attributes
Attributes describe an object's state (`self.name`, `self.age`).

### Methods
A method is a function defined inside a class. It automatically receives the object it was called on as its first parameter, conventionally named `self`.

````python
class Person:
    """Represent a person and provide simple behaviour."""

    def __init__(self, name, age):
        """Initialise the person's name and age."""
        self.name = name
        self.age = age

    def have_birthday(self):
        """Increase the person's age by one year."""
        self.age += 1


person = Person("Amina", 30)
person.have_birthday()
print(person.name, person.age)   # 31
````

### What happens when `person.have_birthday()` runs?

```text
person  ───────►  Person object
                   ┌─────────────┐
                   │ name: Amina │
                   │ age: 30     │
                   └──────┬──────┘
                          │ have_birthday(self=person)
                          ▼
                   age becomes 31
```

Python automatically passes `person` in as `self` -- you never write `self` yourself when *calling* a method, only when *defining* one.

## 5. `__init__()` and object state

`__init__()` is called automatically when an object is created. It is the natural place to validate incoming data and set up initial state.

````python
class BankAccount:
    """Represent a simple bank account."""

    def __init__(self, owner, balance=0):
        """Initialise the account owner and starting balance."""
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        """Add a positive amount to the account."""
        if amount <= 0:
            raise ValueError("Deposit must be positive")
        self.balance += amount

    def withdraw(self, amount):
        """Withdraw money if sufficient funds are available."""
        if amount <= 0:
            raise ValueError("Withdrawal must be positive")
        if amount > self.balance:
            raise ValueError("Insufficient funds")
        self.balance -= amount


account = BankAccount("Amina", 500)
account.deposit(150)
account.withdraw(100)
print(account.owner, account.balance)
````

## 6. Making objects readable with `__str__()`

Without `__str__()`, printing a custom object usually gives a technical representation such as:

```text
<__main__.Person object at 0x...>
```

`__str__()` defines a **human-friendly string representation**. Python uses it whenever `print()` needs to turn your object into text.

````python
class Person:
    """Represent a person with a readable string representation."""

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        """Return a human-readable description of the person."""
        return f"{self.name} (age {self.age})"


person = Person("Amina", 31)
print(person)  # Python calls person.__str__() for us.
````

## 7. Encapsulation: keep data and behaviour together

**Encapsulation** means:

- putting related **attributes and methods** in one class;
- providing a clear interface for working with an object's state;
- reducing unnecessary dependence on the object's internal representation.

Python does not enforce private attributes in the same way as some languages. A common convention is:

- `name` → public attribute
- `_name` → internal/protected-by-convention attribute
- `__name` → name-mangled attribute

**Focus on the design idea:** users of a class should interact through meaningful methods rather than depending on internal details.

````python
class Temperature:
    """Store a temperature in Celsius and expose useful operations."""

    def __init__(self, celsius):
        self._celsius = celsius  # Internal attribute by convention.

    def set_celsius(self, value):
        """Update the temperature after validating the value."""
        if value < -273.15:
            raise ValueError("Temperature cannot be below absolute zero")
        self._celsius = value

    def get_celsius(self):
        """Return the temperature in Celsius."""
        return self._celsius

    def to_fahrenheit(self):
        """Convert the temperature to Fahrenheit."""
        return (self._celsius * 9 / 5) + 32


t = Temperature(20)
print(t.to_fahrenheit())
t.set_celsius(25)
print(t.get_celsius())
````

### Encapsulation diagram

```text
             Temperature object
        +---------------------------+
        | internal state            |
        | _celsius                  |
        +-------------+-------------+
                      |
                 public interface
              +-------+--------+
              |                |
        set_celsius()   to_fahrenheit()
              |                |
              +------ user ---+
```

The user does not need to know how the temperature is stored internally in order to use the class.

## 8. Polymorphism - one interface, different behaviour

Polymorphism means that different objects can respond to the **same method call** in their own way. We can demonstrate this without inheritance -- Python achieves it through **duck typing**: if an object provides the required method, it can be used.

```text
animal.speak()
     |
     +── Dog      → "Woof!"
     +── Cat      → "Meow!"
```

The calling code can use the same interface, `speak()`, without needing to know or check which class it is dealing with.

````python
class Dog:
    """A dog that can speak."""

    def speak(self):
        """Return the sound made by a dog."""
        return "Woof!"


class Cat:
    """A cat that can speak."""

    def speak(self):
        """Return the sound made by a cat."""
        return "Meow!"


def make_it_speak(animal):
    """Use the common speak() interface of the supplied object."""
    print(animal.speak())


animals = [Dog(), Cat()]
for animal in animals:
    make_it_speak(animal)
````

## 9. Putting the pieces together: a small model

Let's combine the main ideas in one class.

**Model:** a `Product` has data and behaviour.

- attributes → name, price, stock
- methods → `restock()`, `sell()`, `__str__()`
- encapsulation → stock changes through methods
- objects → each product has independent state

````python
class Product:
    """Represent a product with price and stock information."""

    def __init__(self, name, price, stock=0):
        """Initialise the product's identity and current stock."""
        self.name = name
        self.price = price
        self.stock = stock

    def restock(self, quantity):
        """Increase stock by a positive quantity."""
        if quantity <= 0:
            raise ValueError("Quantity must be positive")
        self.stock += quantity

    def sell(self, quantity):
        """Reduce stock when enough items are available."""
        if quantity <= 0:
            raise ValueError("Quantity must be positive")
        if quantity > self.stock:
            raise ValueError("Not enough stock")
        self.stock -= quantity

    def __str__(self):
        """Return a concise, human-readable product description."""
        return f"{self.name}: \u00a3{self.price:.2f} | stock={self.stock}"


laptop = Product("Laptop", 899.99, 5)
coffee = Product("Coffee", 3.50, 20)

laptop.sell(1)
coffee.restock(10)

print(laptop)
print(coffee)
````
