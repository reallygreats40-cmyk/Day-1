### The `break` statement
-  terminates the current loop and resumes execution at the next statement

- used mainly when some external condition is triggered requiring a hasty exit from a loop. 

```python
# example 1
for letter in 'Python':     
    if letter == 'h':
        break
    print('Current Letter :', letter)
```

```python
# Example 2
var = 10                  
while var > 0:
    print('Current variable value :', var)
    var = var -1
    if var == 5:
        break

print("Good bye!")
```

### The `continue` statement
- returns control to the beginning of the loop. The continue statement rejects all the remaining statements in the current iteration of the loop and moves the control back to the top of the loop.



```python
# example 1
for letter in 'Python':
    if letter == 'h':
        continue
    print('Current Letter :', letter)
```

```python
# example 2
var = 10                    
while var > 0: 
    var = var -1
    if var == 5:
        continue
    print('Current variable value :', var)
print("Good bye!")
```

#### A function that does not take a parameter and does not return a value

#### A function that takes a parameter, or parameters and returns some result
- example, two params that get multiplied to return a result

```python
def multiply(a, b): # formal parameters
    return a * b

# now call the function
multiply(5,2)
#multiply(int(input("enter a ")), int(input("enter b ")))
```

#### A function with params that have default values
- example, a function to compute area of a circle, pi can be initialised

```python
def compCircleArea(radius, pi = 3.14, k = 1 ):
    return pi * (radius**2)

# now call the function
#compCircleArea(10) 
compCircleArea(10, 3) # replaces 3.14 with 3
```

### Writing to a text file
- open a file (disk) in write mode and get hold of it (memory)
- write to it
- close it

```python
names = ["Andrew", "Jacqui", "Agnes", "Lincoln"] 
age =  [23,54,19,16]

def writeToFile():
    myfile = open("test.txt","w")
    for i in range(0, len(names)):
        myfile.write(names[i]+","+str(age[i])+"\n")
    myfile.close()
    
writeToFile()
```

### Reading from a file
- open the file in read mode 
- read
- close

```python
def readFromFile():
    n = []
    my_file = "test.txt"
    try:
        f = open(my_file, "r")
        for i in f.readlines():
            print(i.strip()) # or try print(i.strip())
            n.append(i)
        print(len(n)) # checking list size has been updated
    finally:
        f.close()
                              

readFromFile()
```

### Read from a CSV file and create a dictionary

```python
def readCsvFileDict():
    with open("staff.csv", 'r') as csvfile:
        reader = csv.DictReader(csvfile)
        for row in reader:
            print(dict(row))
readCsvFileDict()
```

### Create a csv file and write content to it

```python
def writeCsvFile():
    with open('staff.csv', 'w') as staff_file:
        writer = csv.writer(staff_file, delimiter=',')
        writer.writerow(['Name', 'Dept', 'Month'])
        writer.writerow(['James Douglas', 'Engineering', 'November'])
        writer.writerow(['Angel Myers', 'IT', 'March'])
        writer.writerow(['Tabitha', 'Product Design', 'December'])
        staff_file.close()

writeCsvFile()
```

### Pickling objects
- the serializing and de-serializing of python objects to a byte stream. Unpicking is the opposite.
- Pickling is used to store python objects. This means things like lists, dictionaries, class objects, etc

### How to go about pickling
- import the `pickle` module first
- create an object to pickle, e.g, a dictionary
- then open a file (open to write bytes)
- then use `pickle.dump()` to put the dictionary object into the opened file
- then close the file

```python
import pickle

score = {'a':1 , 'b':3, 'c':3, 'd':2, 'e':1, 'f':4, 'g':2, 'h':4, 'i':1, \
         'j':8, 'k':5, 'l':1, 'm':3, 'n':1, 'o':1, 'p':3, 'q':10, 'r':1, 's':1, 't':1, \
         'u':1, 'v':8, 'w':4, 'x':8, 'y':4, 'z':10}


pickle_out = open("myscores.pickle","wb") # we create myscores.pickle on disk
pickle.dump(score, pickle_out)
pickle_out.close()
```

### How do we access the file?
- we unpickle
- Open the pickle file
- Use pickle.load() to load it to a variable.

```python
pickle_in = open("myscores.pickle","rb")
my_dict = pickle.load(pickle_in)

# check out the content
#print(my_dict)
print(my_dict['b'])
```

<b>Pickling is not useful when data has to be shared between Python applications and applications written in other technologies </b>

### coding and decoding json
Python has an built-in module called json for encoding and decoding JSON data.
The process of encoding JSON is  called `serialization`. 
- serialization - transformating  data into a series of bytes.
- deserialization is the reciprocal process of decoding data that has been stored or delivered in the JSON standard.

```python
# let's serialise this
data = {
    "president": {
        "name": "Mike, Michael",
        "species": "Human"
    }
}
```

### we write the above content to a json file
- create a file, e.g.,  `data_file.json` and open it in `write` mode. 
- JSON files have an extension `.json`

```python
import json

with open("data_file.json", "w") as write_file:
    json.dump(data, write_file)
    
```

To continue using the serialized JSON data in your program, you could write it to a native Python `str` object:<br>

```python
json_string = json.dumps(data) # does not write to disk
#print(json_string)
```

### Deserializing JSON
- key functions are `load()` and `loads()` for turning JSON encoded data into Python objects.

```python
with open("data_file.json", "r") as read_file:
    data = json.load(read_file)
    
# object returned is usually a dictionary or a list

#print(type(data))
```

```python
#Suppose you get some data like this from another program
json_string = """
{
    "researcher": {
        "name": "Turing",
        "species": "Human",
        "relatives": [
            {
                "name": "Mathson",
                "species": "Human"
            }
        ]
    }
}
"""

# to deserialise, you do
d = json.loads(json_string)
print(type(d))
print(d)
```

```python
isdir = os.path.isdir(path)
print(isdir)
```

Now use `isfile()` to list files only

```python
from os import listdir
from os.path import isfile, join
onlyfiles = [f for f in listdir(path) if isfile(join(path, f))]

onlyfiles
```

- `os.listdir()` 
    -  lists files and folders in a given directory. 
    - does not return special entries such as `.` and `..` which the operating system uses to navigate through different directories.
    - does not return files and folders beyond the first level of folders. 

### `os.walk()`
- retrieves a list of files contained within a tree. 
- iterates over each directory in a tree. 
- returns the name of every file and folder within a directory and any of its subdirectories.



```python
import os

path = '/Users/skanyaru/Documents/Framework Training/'

for root, directories, files in os.walk(path, topdown=False):
    for name in files:
        print(os.path.join(root, name))
    for name in directories:
        print(os.path.join(root, name))
```

`os.walk()` method accepts four parameters:

- `top`  - the top directory whose component file and folder names you want to retrieve (required)
- `topdown` - when set to True, specifies that directories should be scanned from the top down. If this value is set to False, directories will be scanned from the bottom up (optional)
- `onerror` -  provides an error handler if an error is encountered (optional)
- `followlinks` - if set to True, visits folders referenced by system links (optional)

In the example above, we focussed on the first two parameters. `onerror` and `followlinks` are quite advanced and are rarely  used.

### Closing Files 
- after file performing operations,close the file.

- Closing a file will free up the resources that were tied with the file. It is done using the `close()` method.

- Python has a garbage collector to clean up unreferenced objects but we must not rely on it to close the file.

```python
myfile = open("textfile.txt", 'r')
# perform file operations
myfile.close()
```

Above statements aren't entirely safe. If an exception occurs when we are performing some operation with the file, the code exits without closing the file.

Below is a safer way:

```python
try:
    myfile = open("textfile.txt")
   # perform file operations
finally:
    myfile.close()
```

**Alternative to closing a file**

- best way to close a file is by using the `with` statement. 
-  ensures that the file is closed when the block inside the with statement is exited.

We don't need to explicitly call the `close()` method. It is done internally.

```python
with open("textfile.txt", encoding = 'utf-8') as myfile:
    pass
   # perform file operations
```

### Writing to Files
- open it in write `w`, append `a` or exclusive creation `x` mode.

- `w` mode will overwrite into the file if it already exists. 



```python
with open("test.txt",'w') as myfile:
    myfile.write("my data file\n")
    myfile.write("This file\n\n")
    myfile.write("contains three lines\n")
```

### Reading a File
- open the file in reading `r` mode.

- use the `read(size)` method to read in the `size` number of data. If the size parameter is not specified, it reads and returns up to the end of the file.



```python
myfile = open("test.txt",'r')
myfile.read(4)    # read the first 4 data values
```

```python
myfile.read(4) # read the next 4 data
```

```python
myfile.read()  # read in the rest till end of file
```

```python
myfile.read()  # further reading returns empty sting
```

### Reflections
- the `read()` method returns a newline as '\n'. 
- Once the end of the file is reached, we get an empty string on further reading.

```python
myfile.tell()  # get the current file position
```

```python
print(myfile.read())  # read the entire file
```
