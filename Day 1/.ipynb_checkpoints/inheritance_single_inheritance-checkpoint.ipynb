## 1. A class without inheritance

A class describes the data and behavior its objects share. Here, `Dog` stores a name and can introduce itself.

````python
class Dog:
    def __init__(self, name):
        self.name = name

    def introduce(self):
        return f"I'm {self.name}, a dog."

buddy = Dog("Buddy")
print(buddy.name)
print(buddy.introduce())
````

`__init__` runs when we create an object. The parameter `self` refers to that particular object, so `self.name` stores a value on it.

## 2. What inheritance means

Inheritance lets a new class reuse and specialize another class. The existing class is called the **parent** (or base class); the new class is the **child** (or subclass).

In `class Child(Parent):`, the name in parentheses is the class being inherited from. A child class can use the parent's methods even when it has not defined those methods itself.

````python
class Animal:
    def eat(self):
        return "Eating food"

class Dog(Animal):
    pass

buddy = Dog()
print(buddy.eat())
````

`Dog` inherits from `Animal`, so `Dog` objects can use `eat()`. The `pass` keyword means “no extra code here for now.”

## 3. Inherited methods work with each object's own data

A child class inherits behavior, and its instances can still hold their own attributes. Here, the parent initializes the `name` used by the inherited `introduce()` method.

````python
class Animal:
    def __init__(self, name):
        self.name = name

    def introduce(self):
        return f"I'm {self.name}."

class Dog(Animal):
    pass

class Cat(Animal):
    pass

pets = [Dog("Rex"), Cat("Mittens")]
for pet in pets:
    print(pet.introduce())
````

The `Dog` and `Cat` classes reuse `Animal.__init__` and `Animal.introduce`. Each object has a different `name` value.

## 4. Child classes can add their own behavior

Inheritance does not stop a child from having its own attributes and methods. A `Dog` is an `Animal` with extra dog-specific behavior.

````python
class Animal:
    def __init__(self, name):
        self.name = name

    def introduce(self):
        return f"I'm {self.name}."

class Dog(Animal):
    def fetch(self, item):
        return f"{self.name} fetches the {item}."

buddy = Dog("Buddy")
print(buddy.introduce())  # inherited from Animal
print(buddy.fetch("ball"))  # defined on Dog
````

## 5. Overriding a method

A child can provide its own version of a method that the parent already defines. This is called **overriding**. The child's method is used when called on a child object.

````python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return "Some animal sound"

class Dog(Animal):
    def speak(self):
        return "Woof!"

class Cat(Animal):
    def speak(self):
        return "Meow!"

for animal in [Animal("creature"), Dog("Rex"), Cat("Mittens")]:
    print(animal.name, "says", animal.speak())
````

The parent still provides a general `speak()` behavior. `Dog` and `Cat` replace it with more specific behavior.

## 6. Extending parent behavior with `super()`

Sometimes a child needs to add to the parent's setup rather than replace it. `super()` lets the child call a method from its parent. Use it here to reuse the parent's `__init__` and then store a `breed`.

````python
class Animal:
    def __init__(self, name):
        self.name = name

    def describe(self):
        return f"{self.name} is an animal."

class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)  # let Animal store the name
        self.breed = breed

    def describe(self):
        return f"{super().describe()} Breed: {self.breed}."

buddy = Dog("Buddy", "Labrador")
print(buddy.name)
print(buddy.breed)
print(buddy.describe())
````

`super().__init__(name)` runs the parent's initializer. Without that call, the child initializer would need to set `self.name` itself. In `describe()`, `super().describe()` reuses the parent's description and the child adds more detail.

## 7. Check the inheritance relationship

`isinstance()` can check whether an object belongs to a class or one of its subclasses. `issubclass()` checks the relationship between classes.

````python
class Animal:
    pass

class Dog(Animal):
    pass

buddy = Dog()
print(isinstance(buddy, Dog))
print(isinstance(buddy, Animal))
print(issubclass(Dog, Animal))
````

All three results are `True`: `buddy` is a `Dog`, a `Dog` is also an `Animal` through inheritance, and `Dog` is a subclass of `Animal`.

## 8. When is single inheritance useful?

Use inheritance when the child really is a more specific kind of the parent and can sensibly share some of its behavior. For example, a `Dog` is an `Animal` and can share animal behavior.

If two classes merely need one small helper operation, a standalone function or a composed object may be simpler. Inheritance creates a relationship that future code will rely on, so keep that relationship meaningful.

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def summary(self):
        return f"{self.name} earns £{self.salary} per year."

class Manager(Employee):
    def __init__(self, name, salary, team_size):

print(Manager("Sam", 52000, 6).summary())
```
