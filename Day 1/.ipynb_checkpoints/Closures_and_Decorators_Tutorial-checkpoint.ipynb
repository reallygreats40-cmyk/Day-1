## 1. Functional Programming

Python is a **multi-paradigm** language. One of those paradigms is
*functional programming* — a style built on a single key idea:

> **Functions are first-class citizens.** They can be stored in variables,
> passed to other functions, and returned from functions — exactly like
> integers or strings.

The functional style favours **pure functions** (no side effects, same input
always gives same output) and **composing** small functions into larger ones
rather than mutating shared state.

````python
# Functions are just values — you can assign them to variables
def greet(name):
    return f"Hello, {name}!"

say_hi = greet          # no parentheses — we're referencing the function object
print(say_hi("Alice"))  # Hello, Alice!
print(type(greet))      # <class 'function'>
````

````python
# Functions can be stored in a list and called in a loop
def double(x): return x * 2
def square(x): return x ** 2
def negate(x): return -x

transforms = [double, square, negate]

for fn in transforms:
    print(fn(4))
````

````python
# declarative(functional, logic), imperative (oop, procedural)
````

## 2. Higher-Order Functions

A **higher-order function** is any function that:
- **takes** another function as an argument, **or**
- **returns** a function as its result (or both).

Python's built-ins `map`, `filter`, and `sorted` are classic examples.

````python
numbers = [1, 2, 3, 4, 5]

e = []
for i in numbers:
    e.append(i*2)

e
````

````python
# def product(a, b):
#     return a + b

from functools import reduce

n = reduce(lambda a,b: a + b, numbers) # higher order fncti
n
````

````python
# map() applies a function to every item in an iterable
numbers = [1, 2, 3, 4, 5]
doubled = list(map(double, numbers))
print(doubled)  # [2, 4, 6, 8, 10]
````

````python
# filter() keeps only items for which the function returns True
def is_even(n): return n % 2 == 0

evens = list(filter(is_even, numbers))
print(evens)  # [2, 4]
````

````python
# sorted() accepts a key= function to control sort order
words = ["banana", "fig", "apple", "kiwi"]
print(sorted(words, key=len, reverse = True))  # sort by word length
````

#### You can write your own higher-order function

````python
# Writing your own higher-order function
def apply_twice(fn, value):
    """Apply fn to value, then apply fn to the result."""
    return fn(fn(value)) # what does fn(fn(value)) do? what if it is just fn(value)?

print(apply_twice(double, 3))   # double(double(3)) = double(6) = 12
print(apply_twice(square, 2))   # square(square(2)) = square(4) = 16
````

## 3. Lambda Functions

A `lambda` is an **anonymous function** written in a single expression.
It is ideal for short, throw-away functions passed inline to higher-order
functions.

```
lambda <parameters>: <expression>
```

Lambdas are intentionally limited — one expression only, no statements.
For anything more complex, use a named `def`.

````python
# Equivalent ways to write the same function
def add(a, b): return a + b
add_lambda = lambda a, b: a + b # ideally, you don't want to store it in a reference like this

print(add(3, 4))         # 7
print(add_lambda(3, 4))  # 7
````

````python
# Lambdas shine as inline key functions
students = [("Alice", 88), ("Bob", 72), ("Carol", 95)]

# Sort by score (second element of each tuple)
by_score = sorted(students, key=lambda s: s[1], reverse=True) # sorting descending by value
print(by_score)
````

````python
# Using lambda with map and filter
nums = [1, 2, 3, 4, 5, 6]
odd_squares = list(map(lambda x: x**2, filter(lambda x: x % 2 != 0, nums))) # what does this do?
print(odd_squares)  # [1, 9, 25]
````

## 4. Partial Application

**Partial application** means fixing some of a function's arguments in
advance to produce a new function with fewer parameters.

`functools.partial` does this for you without having to write a wrapper
function by hand.

````python
from functools import partial

def power(base, exponent):
    return base ** exponent

square  = partial(power, exponent=2)   # fix exponent=2
cube    = partial(power, exponent=3)   # fix exponent=3

print(square(5))   # 25
print(cube(3))     # 27
````

````python
# Practical use: pre-configure a print separator
print_bar = partial(print, "-" * 40)
print_bar()
print('something here maybe')
print_bar()
````

````python
# Pre-configure a logger level
import logging
logging.basicConfig(level=logging.DEBUG)
log_info = partial(logging.log, logging.INFO)
log_info("Partial application keeps code DRY.") # DRY - 
````

Every call needs that `logging.INFO` as the first argument. 

`partial(logging.log, logging.INFO)` builds a new function that already has logging.INFO locked in as its first argument.


#### You can also implement `partial` application manually using a closure

````python
def multiplication(a,b,c = 2, d = 5):
    ''' this function takes 4 arguments
    last two have default values of 2 and 5.
    these can be overriden.
    '''
    # this functio
    return a * b * c * d

multiplication(2,3)
help(multiplication) 
````

````python
def make_multiplier(factor):
    """Returns a function that multiplies its argument by factor."""
    def multiply(x):
        return x * factor
    return multiply

triple = make_multiplier(3)
print(triple(7))   # 21
print(triple(10))  # 30
````

## 5. Closures - Functions that Return Functions

The `make_multiplier` example above is a **closure**: an inner function
that *remembers* variables from the enclosing scope, even after the outer
function has returned.

> A closure = inner function + its captured environment.

Think of it as a function carrying a small backpack of values with it.

````python
def make_greeter(greeting):
    """Factory: returns a personalised greeting function."""
    def greet(name):
        # 'greeting' lives in make_greeter's scope, but greet can see it
        return f"{greeting}, {name}!"
    return greet

hello  = make_greeter("Hello")
jambo  = make_greeter("Jambo")   # Swahili greeting

print(hello("Alice"))  # Hello, Alice!
print(jambo("Bob"))    # Jambo, Bob!
````

````python
# Each closure is independent — they don't share state
add5  = make_multiplier(5)
add10 = make_multiplier(10)

print(add5(3))   # 15  — uses factor=5
print(add10(3))  # 30  — uses factor=10
````

## 6. Variable Scoping and Nested Scopes

Python resolves names using the **LEGB** rule - it searches scope in this order:

| Letter | Scope | Example |
|--------|-------|---------|
| **L** | Local | Variables defined inside the current function |
| **E** | Enclosing | Variables in any enclosing (outer) function |
| **G** | Global | Variables defined at module level |
| **B** | Built-in | Python's built-ins (`len`, `print`, …) |

Python looks outward through these layers until it finds the name.

````python
x = "global"          # G — global scope

def outer():
    x = "enclosing"   # E — enclosing scope

    def inner():
        x = "local"   # L — local scope
        print("inner sees:", x)

    inner()
    print("outer sees:", x)

outer()
print("global sees:", x)
````

````python
# If inner does NOT define x, it looks outward (E then G)
y = "global y"

def outer2():
    y = "enclosing y"

    def inner2():
        print("inner2 sees y from enclosing scope:", y)  # 'enclosing y'

    inner2()

outer2()
````

## 7. The Closure Machinery

Python stores captured variables in a special attribute called `__closure__`.
Each captured variable is held in a **cell** object. You can inspect this
directly to see what a closure has captured.

````python
def make_adder(n):
    def add(x):
        return x + n
    return add

add3 = make_adder(3)

# Inspect the closure
print("Closure cells:",  add3.__closure__)
print("Captured value:", add3.__closure__[0].cell_contents)  # 3
print("Free variables:", add3.__code__.co_freevars)           # ('n',)
````

````python
# Multiple captured variables
def make_between_check(lo, hi):
    def between(x):
        return lo <= x <= hi
    return between

is_teen = make_between_check(13, 19)

print(is_teen(15))  # True
print(is_teen(25))  # False

# Both lo and hi are captured
print(is_teen.__code__.co_freevars)  # ('hi', 'lo')
````

## 8. Closures and `nonlocal` Variables

By default, assigning to a variable inside an inner function creates a
**new local** variable - it does not modify the enclosing variable.

Use `nonlocal` to tell Python: *'I want to assign to the variable in the
enclosing scope, not create a new local one.'*

This is essential for closures that need **mutable state**.

````python
# Without nonlocal — count stays 0 forever
def make_broken_counter():
    count = 0
    def increment():
        count = count + 1   # ERROR: UnboundLocalError
        return count
    return increment

# Uncomment to see the error:
#counter = make_broken_counter(); counter()
````

````python
# With nonlocal — works correctly
def make_counter(start=0):
    count = start
    def increment():
        nonlocal count      # 'count' refers to the enclosing variable
        count += 1
        return count
    return increment

counter = make_counter()
print(counter())  # 1
print(counter())  # 2
print(counter())  # 3
````

````python
# Each closure has its own independent counter
counter_b = make_counter(10)
print(counter_b())  # 11
print(counter_b())  # 12
````

## 9. Decorators - Functions that Wrap Functions

A **decorator** is a higher-order function that:
1. Takes a function (`func`) as input
2. Defines a `wrapper` function that adds behaviour around `func`
3. Returns `wrapper`

Decorators are closures applied to the specific problem of extending a
function's behaviour without modifying its source code.

```
╔════════════════════════════╗
║      wrapper function      ║
║  ┌──────────────────────┐  ║
║  │   original function  │  ║
║  └──────────────────────┘  ║
╚════════════════════════════╝
```

````python
# design patterns
````

````python
# The standard decorator shape
def announce(func):              # 1. take a function
    def wrapper(*args, **kwargs):
        print(f"Calling {func.__name__}")
        result = func(*args, **kwargs)  # 2. call original
        print(f"Done")
        return result            # 3. pass the return value through
    return wrapper               # 4. return the wrapper


def add(a, b):
    return a + b

add = announce(add)   # manually apply the decorator
print(add(2, 3))
````

````python
def addition(**args):
    return sum(args)

addition(a = 2,3, 5,9)
````

````python
# Using *args, **kwargs is important — it lets the wrapper accept
# ANY function signature without knowing it in advance

def shout(func):
    def wrapper(*args, **kwargs):
        result = func(*args, **kwargs)
        return result.upper()   # add behaviour to the return value
    return wrapper



# greet = shout(greet)
# print(greet("world"))  # HELLO, WORLD

````

````python
@shout
def greet(name):
    return f"hello, {name}"

greet('john')
````

## 10. Decorator Syntax — `@`

Writing `func = decorator(func)` is the long form. Python provides the
`@decorator_name` shorthand - placed on the line directly above `def`.

These two blocks are **exactly equivalent**:

```python
# Long form
def my_func(): ...
my_func = decorator(my_func)

# Short form
@decorator
def my_func(): ...
```

````python
def shout(func):
    def wrapper(*args, **kwargs):
        result = func(*args, **kwargs)
        return result.upper()   # add behaviour to the return value
    return wrapper


@shout 
def greet(name):
    return f"hello, {name}"


print(greet("world"))  # HELLO, WORLD
````

## 11. Preserving Metadata - `functools.wraps`

When a decorator wraps a function, the wrapper **replaces** the original.
This silently corrupts the function's `__name__`, `__doc__`, and other
attributes - which breaks debugging, help(), and documentation tools.

`@functools.wraps(func)` copies those attributes from `func` onto the
wrapper. **Always include it.**

````python
def my_function():
    """This docstring should be preserved."""
    pass


print(my_function.__name__)  
print(my_function.__doc__) 
````

````python
# Without functools.wraps - identity is lost
def bad_decorator(func):
    def wrapper(*args, **kwargs):
        return func(*args, **kwargs)
    return wrapper

@bad_decorator
def my_function():
    """This docstring should be preserved."""
    pass

print(my_function.__name__)  # 'wrapper'  ← wrong!
print(my_function.__doc__)   # None       ← lost!
````

````python
import functools
````

````python
# With functools.wraps - identity is preserved
def good_decorator(func):
    @functools.wraps(func)   # copies __name__, __doc__, __annotations__ …
    def wrapper(*args, **kwargs):
        return func(*args, **kwargs)
    return wrapper

@good_decorator
def my_function():
    """This docstring should be preserved."""
    pass

print(my_function.__name__)   # 'my_function'  ← correct
print(my_function.__doc__)    # 'This docstring should be preserved.'  ← correct
````

## 12. Decorator Factories - Decorators that Take Arguments

Sometimes you want a decorator to be *configurable*:

```python
@repeat(3)          # run the function 3 times
def say_hi(): ...
```

`repeat(3)` must itself **return** a decorator. So you add one extra layer:

```
decorator_factory(args)
    └─ returns decorator(func)
                └─ returns wrapper(*args, **kwargs)
```

````python
def repeat(x):
    """Decorator factory: repeats the decorated function `x` times."""
    def decorator(func):
        @functools.wraps(func)
        def wrapper(*args, **kwargs):
            for _ in range(x):
                result = func(*args, **kwargs)
            return result
        return wrapper
    return decorator


@repeat(3)
def cheer(name):
    print(f"Hip-hip, hooray for {name}!")

cheer("Alice")
````

## 13. Decorator Order and Built-in Decorators

**Stacking decorators**: multiple `@` lines are applied **bottom-up** - the
decorator closest to the function is applied first.

```python
@A
@B
def f(): ...
# Equivalent to: f = A(B(f))
```

Python's three most important **built-in decorators** are:

| Decorator | Meaning |
|-----------|---------|
| `@staticmethod` | Method with no implicit first argument |
| `@classmethod` | Method receives the class (`cls`) as first argument |
| `@property` | Turns a method into a read-only attribute |

````python
st = 'Python is extensive'

st.upper().split()

@to_upper()
@split_tolist
def say_hello():
    return 'hello there'
````

````python
# Decorator stacking — order matters
def bold(func):
    @functools.wraps(func)
    def wrapper(*a, **k): return f"<b>{func(*a, **k)}</b>"
    return wrapper

def italic(func):
    @functools.wraps(func)
    def wrapper(*a, **k): return f"<i>{func(*a, **k)}</i>"
    return wrapper


@bold          # applied second
@italic        # applied first  → italic(greet_text)
def greet_text(name):
    return f"Hello, {name}"

print(greet_text("Alice"))   # <b><i>Hello, Alice</i></b>
````

````python
# Built-in decorators
class Circle:
    PI = 3.14159

    def __init__(self, radius):
        self._radius = radius

    @property
    def area(self):
        """Accessed like an attribute, not a method call."""
        return self.PI * self._radius ** 2

    @classmethod
    def unit(cls):
        """Alternative constructor: returns a circle of radius 1."""
        return cls(1)

    @staticmethod
    def describe():
        """No access to instance or class needed."""
        return "A circle is a round shape."


c = Circle(5)
print(c.area)             # property — no ()
print(Circle.unit().area) # classmethod as factory
print(Circle.describe())  # staticmethod
````

## 14. Class Decorators

Decorators work on **classes** too. A class decorator receives the class
object, can modify or wrap it, and returns a class (or any callable).

Common uses: adding methods automatically, enforcing constraints, registering
classes in a registry.

````python
# Class decorator that adds a .summary() method to any class
def add_summary(cls):
    def summary(self):
        attrs = ", ".join(f"{k}={v}" for k, v in self.__dict__.items())
        return f"{cls.__name__}({attrs})"
    cls.summary = summary   # attach the new method
    return cls


@add_summary
class Product:
    def __init__(self, name, price):
        self.name  = name
        self.price = price


p = Product("Widget", 9.99)
print(p.summary())
````

## Summary

| Concept | Key idea | One-liner |
|---------|----------|-----------|
| First-class functions | Functions are values | `f = my_func; f()` |
| Higher-order functions | Take or return functions | `map(fn, items)` |
| Lambda | Anonymous single-expression function | `lambda x: x*2` |
| Partial application | Pre-fill some arguments | `partial(pow, exp=2)` |
| Closure | Inner function captures outer variables | `make_adder(3)` |
| LEGB scoping | Local → Enclosing → Global → Built-in | Name lookup order |
| `nonlocal` | Assign to an enclosing variable | `nonlocal count` |
| Decorator | Wraps a function to add behaviour | `def dec(func): …` |
| `@` syntax | Shorthand for `f = dec(f)` | `@dec` above `def` |
| `functools.wraps` | Preserve function identity | Always use it |
| Decorator factory | Decorator that takes arguments | Extra outer layer |
| Stacking | Bottom decorator applied first | `f = A(B(f))` |
| Class decorator | Decorate a class object | `@dec` above `class` |
