````python
class BankAccount:
    """Represent a bank account with basic validation."""

    def __init__(self, balance):
        if balance < 0:
            raise ValueError("Opening balance cannot be negative.")
        self.balance = balance

    def deposit(self, amount):
        """Deposit a positive amount."""
        if amount <= 0:
            raise ValueError("Deposit must be positive.")
        self.balance += amount

    def withdraw(self, amount):
        """Withdraw a positive amount when funds are available."""
        if amount <= 0:
            raise ValueError("Withdrawal must be positive.")
        if amount > self.balance:
            print("Insufficient funds.")
            return False
        self.balance -= amount
        return True

    def get_balance(self):
        """Return the current balance."""
        return self.balance


account1 = BankAccount(100)
account1.deposit(50)
account1.withdraw(75)
print("Balance:", account1.get_balance())

try:
    account1.deposit(-10)
except ValueError as exc:
    print("Validation:", exc)
````

````python
class Item:
    """Represent an asset that can be associated with an Owner."""

    def __init__(self, name, item_id):
        self.name = name
        self.item_id = item_id
        self._owner = None

    def display(self):
        return f"{self.name} ({self.item_id})"

    def rename(self, new_name):
        self.name = new_name

    def get_owner(self):
        return self._owner

    def __str__(self):
        return self.display()


class Owner:
    """Manage a collection of associated Items."""

    def __init__(self, name, owner_id):
        self.name = name
        self.owner_id = owner_id
        self._items = []

    def add_item(self, item):
        if any(existing.item_id == item.item_id for existing in self._items):
            raise ValueError(f"Duplicate item ID: {item.item_id}")

        # If the item belonged to another Owner, detach it first.
        old_owner = item.get_owner()
        if old_owner is not None and old_owner is not self:
            old_owner.remove_item(item)

        self._items.append(item)
        item._owner = self

    def remove_item(self, item):
        if item in self._items:
            self._items.remove(item)
            item._owner = None

    def list_items(self):
        return [item.display() for item in self._items]

    def find_item(self, item_id):
        for item in self._items:
            if item.item_id == item_id:
                return item
        return None

    def create_item(self, name, item_id):
        item = Item(name, item_id)
        self.add_item(item)
        return item


owner1 = Owner("Amina", "O001")
owner2 = Owner("Ben", "O002")

# Composition-style creation: Owner creates the Item.
laptop = owner1.create_item("Laptop", "I001")
phone = owner1.create_item("Phone", "I002")

# Aggregation-style use: Item exists independently first.
tablet = Item("Tablet", "I003")
owner1.add_item(tablet)

print("Owner 1:", owner1.list_items())
print("Search I002:", owner1.find_item("I002"))

# Move an Item from one Owner to another.
owner2.add_item(phone)

print("After moving phone:")
print("Owner 1:", owner1.list_items())
print("Owner 2:", owner2.list_items())
print("Phone owner:", phone.get_owner().name)

# Rename and display.
tablet.rename("Drawing Tablet")
print("Renamed:", tablet)

# Duplicate ID is rejected.
try:
    owner2.add_item(Item("Another Phone", "I002"))
except ValueError as exc:
    print("Validation:", exc)
````
